```python
# Slicing Perform on String

s = "Jayesh Patil"
print(s[:]) # Complete string

print(s[0:6]) # print (Jayesh)
print(s[:-6])# print (Jayesh) using negative indexing

print(s[7:])# print (Patil)
print(s[-5:])# print (Patil) using negative indexing

print(s[:3])# jay
print(s[-12:-9])

print(s[7:10])#pat
print(s[-5:-2])

print(s[2:5]) # 2 to 5
print(s[1:6]) # 1 to 6




```

```python
s = "Jayesh Patil"

print(s[7:]) # 7 to end
print(s[:7]) # begining to index 6
print(s[2:]) # 2 to end

print(s[-3:]) # last 3 characters using negative slicing
print(s[-5:]) # last 5 characters using negative slicing
print(s[-5:]) # patil using neagtive indexes
print(s[-12:-6]) # jayesh using neagtive indexing

print(s[::2]) # print every 2nd character
print(s[1::2])
print(s[::3]) # print every 3rd character



```

```python
s = "Jayesh Patil"
print(s[::-1]) # string in reverse
print(s[-1:-6:-1])# print patil in reverse
print(s[-7:-13:-1])# print Jayesh in reverse
print(s[::2]) # Alternate characters from string
print(s[7::2])# Alternate characters from patil





```

```python
s = "Jayesh Patil"

print(s[1:11]) # Print the string excluding the first and last character
print(s[6:]) # Print the string excluding "Jayesh"
print(s[:7]) #  Print the string excluding "Patil"
print(s[2:9:2]) # Print characters from index 2 to 9 with a step of 2.
print(s[1:10:3]) # Print characters from index 1 to 10 with a step of 3
print(s[::-1]) # Print the whole string using [::-1].

```
