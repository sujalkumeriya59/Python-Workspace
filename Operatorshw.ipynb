```python
# Arithmetic & Comparision operator on  Integer 


# Arithmetic 
a = 10
b = 5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
print(a**b)

# Comparision
print(a == b)   
print(a != b)   
print(a > b)    
print(a < b)    
print(a >= b)   
print(a <= b)
```

```python
# Arithmetic & Comparision operator on Float


#  Arithmetic
a= 15.0
b = 2.5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b) # floor Divison
print(a**b)


# Comparision
print(a == b)   
print(a != b)   
print(a > b)    
print(a < b)    
print(a >= b)   
print(a <= b)

```
